# Python 2:
El programa debe pedir constantemente notas por teclado (entre 0 y 10).

Cuando el usuario escriba "salir", el programa finalizará la introducción de
notas y mostrará:
1. La nota más alta.
2. La nota más baja.
3. La media de todas las notas.
4. El número de aprobados (nota mayor o igual a 5).
5. El número de suspensos (nota inferior a 5).
Si no escribe un numero correcto entre 0 y 10 se indicará con un error y ese número
no se tomará en cuenta
Usa funciones


```
def mostrar_resultados(mayor, menor, suma, contador, suspensos, aprobados):
    print("La nota mayor es:", mayor)
    print("La nota menor es:", menor)

    if contador > 0:
        media = suma / contador
        print("La media de las notas es:", media)
    else:
        print("La media de las notas es: 0")

    print("El número de suspensos es:", suspensos)
    print("El número de aprobados es:", aprobados)


mayor = None
menor = None
suma = 0
contador = 0
numSuspensos = 0
numAprobados = 0

while True:
    entrada = input("Ingrese una nota entre 0 y 10 ")

    if entrada == "salir":
        mostrar_resultados(mayor, menor, suma, contador, numSuspensos, numAprobados)
        break

    nota = int(entrada)

    if nota < 0 or nota > 10:
        print("La nota debe estar entre 0 y 10.")
        continue

    if mayor is None or nota > mayor:
        mayor = nota
    if menor is None or nota < menor:
        menor = nota

    suma += nota
    contador += 1

    if nota < 5:
        numSuspensos += 1
    else:
        numAprobados += 1

    print("Nota registrada:", nota)1
```

# Python 3:
Crea un programa en Python que permita al usuario introducir números enteros por teclado y almacenarlos en una lista (array).

El programa debe pedir números constantemente hasta que el usuario escriba "salir".

Una vez finalizada la introducción de números, el programa debe mostrar:
1. La lista completa de números introducidos.
2. La cantidad de números almacenados.
3. La suma de todos los números.
4. El número mayor y el número menor.
5. Una nueva lista que contenga únicamente los números pares.
6. La lista original ordenada de menor a mayor.

Es obligatorio usar listas

```
def mostrar_resultados(numeros, cantNumeros):

    suma = sum(numeros)

    numero_mayor = max(numeros)
    numero_menor = min(numeros)

    for num in numeros :
            if num % 2 == 0:
                numeros_pares.append(num)

    numeros_ordenados = sorted(numeros)

    print("Lista de números :", numeros)
    print("Cantidad de números almacenados:", cantNumeros)
    print("Suma de todos los números:", suma)
    print("Número mayor:", numero_mayor)
    print("Número menor:", numero_menor)
    print("Números pares:", numeros_pares)
    print("Lista ordenada de menor a mayor:", numeros_ordenados)


numeros_pares = []
numeros = []
cantNumeros = 0

while True:
    entrada = input("Ingrese un número entero: ")

    if entrada == "salir":
        mostrar_resultados(numeros, cantNumeros)
        break

    numero = int(entrada)
    numeros.append(numero)
    cantNumeros += 1
```
